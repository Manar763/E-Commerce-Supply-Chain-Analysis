```python
## Fix the 3 shifted rows
## In these rows, the values are in the wrong columns:
## City has the state, State has the zipcode, Street has the city
bad_rows = df_customers['Customer State'].str.isdigit()

## Save the correct values first
correct_city = df_customers.loc[bad_rows, 'Customer Street']
correct_state = df_customers.loc[bad_rows, 'Customer City']
correct_zip = df_customers.loc[bad_rows, 'Customer State'].astype(int)

## Put each value in the right column
df_customers.loc[bad_rows, 'Customer City'] = correct_city
df_customers.loc[bad_rows, 'Customer State'] = correct_state
df_customers.loc[bad_rows, 'Customer Zipcode'] = correct_zip
df_customers.loc[bad_rows, 'Customer Street'] = 'Unknown'
```

```python
## Add column: Customer Full Name = First name + space + Last name
df_customers['Customer Full Name'] = (df_customers['Customer Fname'] + ' ' + df_customers['Customer Lname']).str.strip()
```

##### Edit order date Type

```python
## Edit The date Type
is_number_order = df_orders['order date (DateOrders)'].str.fullmatch(r'\d+(\.\d+)?')
print('Excel numbers :', is_number_order.sum(), ' Text dates :', (~is_number_order).sum())
df_orders.loc[is_number_order, 'order date (DateOrders)'].head()
```

```python
##  Step 1 : text dates -> real dates (the text format is month/day/year)
text_dates = pd.to_datetime(df_orders.loc[~is_number_order, 'order date (DateOrders)'], format = '%m/%d/%Y %H:%M')

## Step 2 : Excel numbers -> real dates (Excel day 0 = 1899-12-30), rounded to the minute
excel_dates = pd.to_datetime('1899-12-30') + pd.to_timedelta(df_orders.loc[is_number_order, 'order date (DateOrders)'].astype(float), unit = 'D')
excel_dates = excel_dates.dt.round('min')

## Step 3 : in the Excel dates the DAY and the MONTH are swapped (ex: 3 Feb 2018 was saved as 2 Mar 2018 = 43161). Swap them back
excel_dates = pd.to_datetime(dict(year = excel_dates.dt.year, month = excel_dates.dt.day, day = excel_dates.dt.month, hour = excel_dates.dt.hour, minute = excel_dates.dt.minute))

## Step 4 : put the 2 parts back in the table (the index keeps the same row order)
df_orders['order date (DateOrders)'] = pd.concat([text_dates, excel_dates]).sort_index()
```

##### Edit Shipping date Type

```python
## [EDIT] Same 4 steps for the shipping date (it has the same 2 formats)
is_number_ship = df_orders['shipping date (DateOrders)'].str.fullmatch(r'\d+(\.\d+)?')
print('Excel numbers :', is_number_ship.sum(), ' Text dates :', (~is_number_ship).sum())

text_dates = pd.to_datetime(df_orders.loc[~is_number_ship, 'shipping date (DateOrders)'], format = '%m/%d/%Y %H:%M')

excel_dates = pd.to_datetime('1899-12-30') + pd.to_timedelta(df_orders.loc[is_number_ship, 'shipping date (DateOrders)'].astype(float), unit = 'D')
excel_dates = excel_dates.dt.round('min')

excel_dates = pd.to_datetime(dict(year = excel_dates.dt.year, month = excel_dates.dt.day, day = excel_dates.dt.month, hour = excel_dates.dt.hour, minute = excel_dates.dt.minute))

df_orders['shipping date (DateOrders)'] = pd.concat([text_dates, excel_dates]).sort_index()
```
